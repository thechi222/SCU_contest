# 影像放大快速開始

把圖片放到 `/work`,權重已掛載於 `/models`。
可自行調整倍率、tile 大小或批次處理多張圖片。


```python
from realesrgan import RealESRGANer
from basicsr.archs.rrdbnet_arch import RRDBNet
import cv2

model = RRDBNet(num_in_ch=3, num_out_ch=3, num_feat=64, num_block=23, num_grow_ch=32, scale=2)
upsampler = RealESRGANer(scale=2, model_path='/models/RealESRGAN_x2plus.pth',
                         model=model, half=True)
image = cv2.imread('/work/sample.png', cv2.IMREAD_UNCHANGED)
output, _ = upsampler.enhance(image, outscale=2)
cv2.imwrite('/work/upscaled.png', output)
```
