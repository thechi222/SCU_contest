# 語音轉逐字稿快速開始

把音訊檔放到 `/work`,模型權重已掛載於 `/models`。
可自行調整語言、beam size 或輸出格式。


```python
from faster_whisper import WhisperModel

model = WhisperModel('/models/whisper-small', device='cuda', compute_type='float16')
segments, info = model.transcribe('/work/sample.wav', language='zh', beam_size=5)
for segment in segments:
    print(f'[{segment.start:.1f}-{segment.end:.1f}] {segment.text}')
```
